```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_dataset.csv")

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
df.head()
```

```python
print("Number of rows and columns:")
print(df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
print("Data types:")
print(df.dtypes)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df.columns.tolist())
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
text_columns = [
    "director",
    "cast",
    "country",
    "date_added",
    "rating"
]

for column in text_columns:
    df[column] = df[column].fillna("Unknown")

print("Missing text values handled successfully.")
```

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicates after removal:")
print(df.duplicated().sum())
```

```python
df["type"] = (
    df["type"]
    .astype(str)
    .str.strip()
    .str.title()
)

print("Type values:")
print(df["type"].unique())
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)

print("Date data type:")
print(df["date_added"].dtype)
```

```python
df["duration"] = (
    df["duration"]
    .astype(str)
    .str.strip()
)

print("Duration cleaned successfully.")
```

```python
print("Content types:")
print(df["type"].value_counts())

print("\nRatings:")
print(df["rating"].value_counts().head(10))
```

```python
print("FINAL DATA QUALITY CHECK")
print("------------------------")

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDataset shape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
